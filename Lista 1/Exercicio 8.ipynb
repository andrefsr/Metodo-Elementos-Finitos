Letra a)

```python
import numpy as np
import matplotlib.pyplot as plt

xi = np.linspace(-1, 1, 100)
eta = np.linspace(-1, 1, 100)

XI, ETA = np.meshgrid(xi, eta)

L1_xi = 0.5 * XI * (XI - 1)
L2_xi = 1 - XI**2
L3_xi = 0.5 * XI * (XI + 1)

L1_eta = 0.5 * ETA * (ETA - 1)
L2_eta = 1 - ETA**2
L3_eta = 0.5 * ETA * (ETA + 1)

N1 = L1_xi * L1_eta
N2 = L2_xi * L1_eta
N3 = L3_xi * L1_eta
N4 = L1_xi * L2_eta
N5 = L2_xi * L2_eta
N6 = L3_xi * L2_eta
N7 = L1_xi * L3_eta
N8 = L2_xi * L3_eta
N9 = L3_xi * L3_eta

N = [N1, N2, N3, N4, N5, N6, N7, N8, N9]

fig = plt.figure(figsize=(15, 12))

for i in range(9):
    ax = fig.add_subplot(3, 3, i+1, projection='3d')
    ax.plot_surface(XI,ETA,N[i],cmap='viridis')
    ax.set_title(f'$N_{i+1}(\\xi,\\eta)$')
    ax.set_xlabel(r'$\xi$')
    ax.set_ylabel(r'$\eta$')
    ax.set_zlabel(r'$N$')

plt.tight_layout()
plt.show()
```

Letra b)

```python
x = np.linspace(-1, 1, 200)
y = np.linspace(-1, 1, 200)

X, Y = np.meshgrid(x, y)

def g(x):
    return np.where(x <= 0,(2*x + 1)*(x + 1),(2*x - 1)*(x - 1))

Phi = g(X) * g(Y)

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Phi)
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_zlabel(r'$\phi(x,y)$')
ax.set_title('Função de forma global associada ao nó central')
plt.show()
```
