```python
import numpy as np
import matplotlib.pyplot as plt

n = [7,10,20]
da = 0 
db = 1
xa = 0.2
xb = 0.3

def f(xm,xa,xb):
    if (xm <= xb) & (xm >= xa): return 1
    else: return 0

plt.figure()
plt.grid(alpha=0.3)
plt.xlabel('x')
plt.ylabel('u(x)')

for j in range(3):
    delta = (db-da)/n[j]
    x = np.linspace(da,db,n[j]+1)
    xm = np.linspace(da+delta/2, db-delta/2, n[j])
    A = np.zeros((n[j],n[j]))
    b = np.zeros(n[j])

    for i in range(n[j]):
        ii = i + 1
        b[i] = ((xb**2 - xa**2) / 2- (xb**(ii + 2) - xa**(ii + 2))/ (ii + 2))

    for i in range(n[j]):
        ii = i + 1
        for k in range(n[j]):
            jj = k + 1
            A[i, k] = ((ii + 1) * (jj + 1)/ (ii + jj + 1) - 1)

    alpha = np.linalg.solve(A,b)

    u = np.zeros(n[j]+1)
    for i in range(1,n[j]+1):
        u[i] = np.sum(alpha * (x[i] - x[i]**(np.arange(1,n[j]+1)+1)))

    plt.plot(x,u,'o',ls='-',label=f'n = {n[j]}')

plt.title('Galerkin Method')
plt.legend()
plt.show()
```
